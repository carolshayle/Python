```python
from google.colab import files
import os

print("Upload your District Shapefile (select .shp, .shx, .dbf, .prj) and WorldPop .tif")
uploaded = files.upload()

# Identify the file names
shp_file = [f for f in uploaded.keys() if f.endswith('.shp')][0]
tif_file = [f for f in uploaded.keys() if f.endswith('.tif')][0]

print(f"Using Shapefile: {shp_file}")
print(f"Using Population Raster: {tif_file}")
```

```python
import geopandas as gpd
import pandas as pd
import numpy as np
import rasterio
from rasterio.mask import mask
from sklearn.cluster import KMeans
from shapely.geometry import Point, MultiPoint
from shapely.ops import voronoi_diagram
```

```python
# 1. Load Data
districts = gpd.read_file(shp_file)
if districts.crs != "EPSG:4326":
    districts = districts.to_crs("EPSG:4326")

# We need to project to a metric system to calculate area accurately in km2
# UTM Zone 38N (EPSG:32638) is good for Somalia
districts_metric = districts.to_crs(epsg=32638)
districts['area_km2'] = districts_metric.area / 10**6

name_col = districts.columns[0] # Assuming first column is the name

def get_num_clusters(area):
    if area < 2000: return 3
    elif area < 5000: return 4
    elif area < 10000: return 5
    else: return 6

def generate_balanced_localities(district_row, raster_path):
    geom = district_row.geometry
    area = district_row['area_km2']
    num_clusters = get_num_clusters(area)

    with rasterio.open(raster_path) as src:
        try:
            # Mask raster to district
            out_image, out_transform = mask(src, [geom], crop=True)
            data = out_image[0]

            # Find populated pixels
            valid_mask = (data > 0)
            values = data[valid_mask]
            rows, cols = np.where(valid_mask)

            # If the district is unpopulated in the data, use geometric centers
            if len(rows) < num_clusters:
                # Fallback: Generate points based on geometry, not population
                points = np.array([[geom.centroid.x, geom.centroid.y]])
                # (Simplification: In real usage, we'd sample random points in the polygon)
                num_clusters = 1 # Keep as one if no data
            else:
                coords = []
                for r, c in zip(rows, cols):
                    coords.append(rasterio.transform.xy(out_transform, r, c))
                points = np.array(coords)

            # Weighted K-Means based on 2026 population counts
            kmeans = KMeans(n_clusters=num_clusters, n_init=10, random_state=42)
            kmeans.fit(points, sample_weight=values)
            centers = kmeans.cluster_centers_

            # Generate Voronoi
            poly_centers = MultiPoint([Point(c) for c in centers])
            vor_regions = voronoi_diagram(poly_centers, envelope=geom.buffer(0.2))

            # Convert to GDF and clip
            localities = gpd.GeoDataFrame(geometry=list(vor_regions.geoms), crs="EPSG:4326")
            clipped = gpd.clip(localities, geom)
            return clipped

        except Exception as e:
            print(f"Skipping {district_row[name_col]} due to error: {e}")
            return None

# 2. Iterate through districts
all_localities = []
for idx, row in districts.iterrows():
    print(f"Processing {row[name_col]} (Area: {row['area_km2']:.1f} km2)...")
    res = generate_balanced_localities(row, tif_file)
    if res is not None:
        res['parent_dist'] = row[name_col]
        all_localities.append(res)

# 3. Finalize Dataset
final_gdf = pd.concat(all_localities).reset_index(drop=True)

# Clean up geometry (smooth the Voronoi lines)
final_gdf['geometry'] = final_gdf.geometry.buffer(0.001, join_style=1).buffer(-0.001, join_style=1)

# Generate Addressing IDs
final_gdf['loc_id'] = final_gdf.groupby('parent_dist').cumcount() + 1
final_gdf['address_id'] = (
    final_gdf['parent_dist'].str[:3].str.upper() +
    "-" +
    final_gdf['loc_id'].astype(str).str.zfill(2)
)

# 4. Export
output_file = "Somalia_Admin3_Addressing_System.geojson"
final_gdf.to_file(output_file, driver='GeoJSON')
files.download(output_file)

print("Finished! Localities sized by Area and centered by 2026 Population.")
```
